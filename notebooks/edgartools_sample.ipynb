# edgartools sample

Minimal SEC EDGAR pulls via [edgartools](https://github.com/dgunning/edgartools). Uses `EDGAR_IDENTITY` from repo `.env`.

```python
from pathlib import Path
import os

from edgar import Company, set_identity

# load .env without adding python-dotenv
env = Path("../.env")
for line in env.read_text().splitlines():
    if line.startswith("EDGAR_IDENTITY="):
        os.environ["EDGAR_IDENTITY"] = line.split("=", 1)[1].strip()
        break

set_identity(os.environ["EDGAR_IDENTITY"])
print("identity:", os.environ["EDGAR_IDENTITY"])
```

```python
co = Company("AAPL")
print(co.name, co.cik)

filing = co.latest("8-K")
filing
```

```python
co.income_statement(periods=3)
```

```python
tenk = filing.obj()
print(tenk.date_of_report)
```
